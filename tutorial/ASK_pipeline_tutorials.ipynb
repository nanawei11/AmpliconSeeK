# ASK pipeline tutorials

This notebook is a simple click-to-run guide for AmpliconSeeK (ASK).

Choose one of the three routes below:

1. Start from an NCBI/SRA accession.
2. Start from FASTQ files.
3. Start from an existing BAM file.

Most users only need to edit the paths in the configuration cell and then run the next cell.


## Setup

Run this notebook from the AmpliconSeeK repository root. The helper functions are loaded from `pipeline/ask_pipeline.py`.


````python
from pathlib import Path
import sys

PROJECT_DIR = Path.cwd()
sys.path.insert(0, str(PROJECT_DIR / "pipeline"))

from ask_pipeline import run_from_sra, run_from_fastq, run_from_bam

print("Project:", PROJECT_DIR)
print("Pipeline helper loaded")
````

## Option 1: Start from NCBI/SRA

Edit the sample name, SRR accession, and BWA index path.

Set `dry_run = True` to print commands without running them. Set it to `False` when the paths are ready.


````python
sample_id = "GBM_sample"
srr_ids = "SRR9140739"
bwa_index = "/path/to/hg38_bwa_index.fa"
workdir = "analysis_from_sra"
genome = "hg38"
threads = 5
dry_run = True
````

````python
run_from_sra(
    sample_id=sample_id,
    srr_ids=srr_ids,
    bwa_index=bwa_index,
    workdir=workdir,
    genome=genome,
    threads=threads,
    dry_run=dry_run,
)
````

## Option 2: Start from FASTQ

Use this when FASTQ files are already available. For single-end data, set `fastq2 = None`.


````python
sample_id = "GBM_sample"
fastq1 = "/path/to/sample_R1.fastq.gz"
fastq2 = "/path/to/sample_R2.fastq.gz"
bwa_index = "/path/to/hg38_bwa_index.fa"
workdir = "analysis_from_fastq"
genome = "hg38"
threads = 5
dry_run = True
````

````python
run_from_fastq(
    sample_id=sample_id,
    fastq1=fastq1,
    fastq2=fastq2,
    bwa_index=bwa_index,
    workdir=workdir,
    genome=genome,
    threads=threads,
    dry_run=dry_run,
)
````

## Option 3: Start from BAM

Use this when the BAM is already sorted and indexed. The `.bai` file should be in the same folder as the BAM.


````python
sample_id = "testdata"
bam = "exampledata/testdata.bam"
workdir = "analysis_from_bam"
genome = "hg38"
dry_run = True
````

````python
run_from_bam(
    sample_id=sample_id,
    bam=bam,
    workdir=workdir,
    genome=genome,
    dry_run=dry_run,
)
````

## Expected outputs

For a sample named `testdata`, ASK writes results under:

```text
analysis_from_bam/testdata/ask/
```

Important files include:

```text
testdata_ask_amplicon_circular_new.tsv
testdata_ask_amplicon_circular_stat_new.tsv
testdata_ask_breakpoint_pair.tsv
testdata_ask_plot/
```

The example BAM should produce a circular amplicon plot similar to:

```text
analysis_from_bam/testdata/ask/testdata_ask_plot/circular_circ_6.pdf
```
